```python
a1=np.array([1, 2, 3, 4, 5])
a2=np.array([6, 7, 8, 9, 10])
print("Addition:", a1 + a2)
print("Subtraction:", a1 - a2)
print("Multiplication:", a1 * a2)
print("Division:", a1 / a2)
print("Exponentiation:", a1 ** 2)
print("Square Root of a1:", np.sqrt(a1))
print("Square Root of a2:", np.sqrt(a2))
print("Maximum:", np.max(a1))
print("Minimum:", np.min(a1))
print("Average:", np.mean(a1))
print("Standard Deviation:", np.std(a1))
print("Variance:", np.var(a1))
print("length:", len(a1))
print("log:", np.log(a1))
print("Aabsolute:", np.abs(a1))
print("Sum:", np.sum(a1))
print("Sum:", np.sum(a2))
#Scalar Operations
print(a1 + 5)
print(a1 * 2)
print(a1 - 3)
print(a2 / 2)
print(a2 ** 2)
#comparison operator
print(a1 > 3)
print(a1[a1 > 3])
#concatenation
new = np.concatenate((a1, a2))
print("Concatenated Array:")
print(new)
```
